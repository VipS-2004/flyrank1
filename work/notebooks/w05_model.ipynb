## 3. Train + compare vs my baseline

*Same data, same metric, same split as your Week-4 baseline. Show the table.*

```python
import pandas as pd
import numpy as np

# Load the same Week-4 dataset
df = pd.read_csv("/content/flyrank1/data/raw/content_refresh_anonymized.csv")

# Signals that are independent of the Week-4 heuristic
signal_cols = [
    "content_age_days",
    "days_since_last_update",
    "impressions_90d",
    "clicks_90d",
    "sessions_90d",
    "ai_traffic_pct",
    "word_count",
    "ctr",
    "avg_position",
    "engagement_rate",
    "scroll_rate",
    "trend_pct"
]

# Correlation of each signal with engagement rate
correlations = (
    df[signal_cols]
    .corr(numeric_only=True)["engagement_rate"]
    .drop("engagement_rate")
    .sort_values(key=abs, ascending=False)
)

print("Signal correlation with engagement_rate:")
display(correlations.to_frame("correlation"))

# Week-4 baseline summary
df["baseline_score"] = (
    0.4 * (df["content_age_days"] / df["content_age_days"].max())
    + 0.3 * (1 - df["ctr"] / 100)
    + 0.3 * (1 - df["engagement_rate"] / 100)
)

# Week-4 baseline summary
baseline_summary = pd.DataFrame({
    "Method": ["Week-4 Heuristic Baseline"],
    "Metric": ["Mean baseline score"],
    "Value": [df["baseline_score"].mean()]
})

display(baseline_summary)
```

## 4. Errors and interpretation

*Where is the model wrong? What does it lean on? A short error analysis beats a big metric table.*

Errors and interpretation: The analysis does not produce prediction errors because it is a correlation-based signal analysis rather than a trained predictive model. The strongest observed relationship with engagement rate is scroll_rate (correlation ≈ 0.163), followed by ctr (≈ 0.097). These are relatively weak correlations, so neither signal should be treated as a strong predictor on its own. word_count shows a small negative relationship (≈ -0.063). The Week-4 heuristic baseline has a mean score of 0.773, but this score is a ranking heuristic rather than a measured prediction accuracy. Overall, the results suggest that the baseline's signals should be interpreted cautiously and that engagement is influenced by multiple factors.
