## Data Limits

This dataset contains historical search and engagement information only.

It cannot explain external business events, marketing campaigns, seasonal effects, or manual content improvements.

The recommendations produced from this dataset should support human decision-making rather than replace it.

```python
df["leak_feature"] = (df["trend_direction"]=="down").astype(int)

print(df[["trend_direction","leak_feature"]].head())

df.drop(columns=["leak_feature"], inplace=True)

print("\nLeakage feature removed successfully.")
```
